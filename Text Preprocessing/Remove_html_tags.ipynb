## Remove HTML Tags

## Library Used:
re - Regular Expression (inbuilt)
to remove html tags by regex

```python
#re - regulal expression
import re 
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
text = "<html><body><p> Movie 1</p> Actor - Ranbeer Kapoor</p><p> Click here to <a href = 'http://goggle.com'> download</a></p></body><html>"
```

```python
remove_html_tags(text)
```

## Remove html tag
HTML tags are written inside < > brackets.
Removing them means extracting only plain 

## why to Remove HTML Tags?
In web scraping, the data is received in HTML format.
For NLP / Text analysis, only clean and plain text is required.Links, tags and markup are considered as noise and need to be removed
