## Regular Expression

```python
import re 
sent3 = 'I am going to delhi'
tokens = re.findall("[\w']+",sent3)
tokens
```

```python
text = """Lorem Ipsum is simply dumpy test of the printing and typesetting industry?
Lorem Ipsum has been the industry's standard dumy text ever since the 1500s. when an unkown printer took a galley of type and scrambled it is to make a type specism book."""
sentences = re.compile('[.!?]').split(text)
sentences
```

## NLTK

```python
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
sent1 = 'I am going to delhi' 
word_tokenize(sent1)
```

```python
text = """Lorem Ipsum is simply dumpy test of the printing and typesetting industry?
Lorem Ipsum has been the industry's standard dumy text ever since the 1500s. when an unkown printer took a galley of type and scrambled it is to make a type specism book."""
sent_tokenize(text)
```
