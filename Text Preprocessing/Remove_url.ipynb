## URL (Uniform Resource Locator) is a web address like http://google.com or www.example.com.

```python
import re
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\s+')
    return pattern.sub(r'', text)
```

```python
text1 = 'check out https://dl.flipkart.com/s/ng!1g4NNNN'
text2 = 'Google search here www.google.com'
```

```python
text2
```

```python
text1
```

```python
remove_url(text1)
```

## why to Remove URL?
For text cleaning in NLP.URLs do not add meaning to sentiment analysis.They are considered as noise.

##  Pattern Used:
r'https?://\S+|www\.\S+'
