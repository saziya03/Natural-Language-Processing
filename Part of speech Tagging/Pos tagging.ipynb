```python
import spacy 
```

### Load english lang dict

```python
nlp = spacy.load('en_core_web_sm')
```

### Create doc 


```python
doc = nlp(u"I will google about facebook")
```

```python
spacy.explain('ADP')
```

```python
for word in doc:
    print(word.text, "-------->", word.pos_, word.tag_,spacy.explain(word.tag))
```
